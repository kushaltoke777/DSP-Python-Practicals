```python
a, b, c = map(int, input("Enter value for a,b,c:-").split())

numbers = [a, b, c]

greatest = [x for x in numbers if x >= a and x >= b and x >= c][0]

print("The greatest number is:", greatest)
```

```python
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Elements starting from index 2
b = a[2:]
print(b)

# Elements from index 0 to 3
c = a[:3]
print(c)
```

```python
a = [1, 2, 3, 4, 5, 6, 7, 8, 9]

# Last two elements
b = a[-2:]
print(b)

# Elements from index 0 to 5
c = a[0:6]
print(c)

# Elements from index -4 to -1
d = a[-4:-1]
print(d)
```

```python
months = ['January', 'February', 'March']

prices = [238.11, 237.81, 238.91]

# Identify minimum price
min_price = min(prices)
print(min_price)

# Identify minimum price index
min_index = prices.index(min_price)
print(min_index)

# Identify month with minimum price
min_month = months[min_index]
print(min_month)
```
